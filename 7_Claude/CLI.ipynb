# 1. Harness Engineering

```python
Evolution of Harness:
1. Prompt Enginnering: 
    - Simple prompting ChatGPT was not good enough to get the work done.
    - This reminds the agent, who the agent, give the agent a persona and a task.
    - Initially it had 4096 tokens, so how to recycle this memory space to get more work done, so we got context eng.
    - With just a prompt, we have so much of context left empty, so better to give more information.
    - For Example, any prompt which does not require any external tool call.
        
2. Context Engineering:
    - Context Eng includes:
        - Tool calling to expore repo and docs.
        - MCP: vendor specific fuctions.
        - RAG: external DB call to get the data.
    - But, long running tasks were not effective. As context window fills, it does the context summarization which looses a lot of information.
    - So, we need a system to better manage the context. So, people started with sub-agents.
    - Example: prompt that requires web search or any tool call.
        
3. Harness Engineering:
    - Harness manages the agent context from outside. It break out the long running tasks into steps, and invoke independent agents to do so.
    - To get Better orchestration, better execution, better context management are required for Harness eng.
    - It uses both Prompt and Context Engineering.
    - It runs on loop/ iteration, which each iteration it gives a fresh context and prompt.
    - Example Prompt: Clone NASA website.
        - Now, this require multiple agents to check the website and some coding agnets to code it. It is a long running task, so the manager agents manages everthing.
        
4. Loop Engineering:
    - Create a schedule, that calls the harness agent in a loop.
```

```python
It means designing the environment around an AI agent so it can do useful work reliabily.
It includes instructions, memory, tools access, tests, permissions, feedback, project rules, workflows, Validation, Automation.

    
For a coding project:
1. Instructions:	A short AGENTS.md explaining project conventions and where to find details.
2. Context:	Accessible architecture docs, requirements, and examples.
3. Tools:	Reliable ways to start the app, inspect the browser, and read logs.
4. Checks:	Tests, linters, and CI that detect mistakes automatically.
5. Boundaries:	Clear limits on file access, commands, and actions requiring approval.
6. Feedback:	Useful failure messages that help the agent diagnose and correct problems.

Steps inside AI harness:
User Request > Planning > Context Collection > Tool Execution > Validation
 
Requirements:
1. Design Workflow: AI-driven development pipelines
2. Build Harness: Confirgure agents with context, rules, and permissions
3. Validate Output: Review AI-generated changes
4. Automate Pipelines: agent powered automation

```
