# 1. Commands

```python
1. /compact: Compact and clear current session context history. will compact the conversation: It will generate a summary of the conversation, including some instructions (for itself) on how to proceed once the compaction process is over.
2. /clear: Clear session context window. Similar to starting a new session of claude code, It clears the context window.
3. /resume: To jump/continue from any previous sessions.
4. /sandbox: with the mode (with auto-allow), Auto-allow mode: will attempt to run inside the sandbox and are automatically allowed without requiring permission. Commands that cannot be sandboxed, such as those needing network access to non-allowed hosts, fall back to the regular permission flow.
5. /help: List available commands and get usage help.
6. /model: Choose active model for current session.
7. /config: Open interactive settings menu.
8. /context: View statistics about current context window & usage.
9. /usage: View current usage for active plan.
10. /init: Analyze project and create initial CLAUDE.MD file.
11. /mcp: View and manage installed MCP servers.
12. /permissions: View and update / change permissions.
13. /rewind: Rewind (undo) to earlier point, same as ESC + ESC.
14. /statusline: Configure the Claude Code status line.
15. /teleport: Resume a remote Claude Code session.    
16. /memory: You can disable / enable and configure this Auto Memory feature via the /memory command inside of Claude Code. This command also allows you to browse and edit the memory files belonging to your project (if any exist).
```

```python
# CLI commands
1. claude or claude 'my prompt': Start a new session, optionally with an initial prompt.
2. claude -p "my prompt": Query with "my prompt" and quit (no ongoing session).
3. claude -c: Continue the last session
4. claude --agent DocsExplorer: Start session with a custom agent.
5. claude --allowedTools "Read" "Write": Skip permission confirmation for specified tools.
6. claude --disallowedTools "Write": Disallow certain tools from context.
6. claude --dangerously-skip-permissions: This will have all permission to edit files as well as to run any terminal commands. This can erase your drive as well.
7. claude --append-system-prompt"..": Append instruction to default system prompt.
8. claude --model opus: Set default model for current session.
9. claude --permission-mode plan: Start in plan mode. This does not have permissions to edit files.
10. claude --remote "Add dark mode": Start a remote (web) session.
11. claude --system-prompt '...': Replace entire default system prompt.

# Add MCP server
1. claude mcp add context - nx -y @upstash/context7-mcp: Add server specific to that project
2. claude mcp add context7 --scope user -- npx -y @upstash/context7-mcp: Add server to all projects
```

# 9. Hooks

```python
Hooks fire at specific points during a Claude Code session. When an event fires and a matcher matches, Claude Code passes JSON context about the event to your hook handler. For command hooks, input arrives on stdin. For HTTP hooks, it arrives as the POST request body. Your handler can then inspect the input, take action, and optionally return a decision. Events fall into three cadences: once per session (SessionStart, SessionEnd), once per turn (UserPromptSubmit, Stop, StopFailure), and on every tool call inside the agentic loop (PreToolUse, PostToolUse):

{ 
    "hooks": {
    "PostToolUse": [
        {
            "matcher"; "Edit |Write",
            "hooks": [
                    {
                        "type": "command",
                        "command": "cd \"$CLAUDE_PROJECT_DIR\" && bun run format 2>/dev/null || true"
                    }
            ]
        }
}
```

# 8. Prompts

```python
1. Concise & Precise: Keep your prompts concise and focused.
2. No Unnecessary Context: Providing useful (!) context is crucial.   Add file reference @file.md that contains important information.
3. Think, Plan, Prompt
    - When working with AI, you may be tempted to just start typing and then "fix stuff over time". Don't do that!
    - Think first, then make a plan, then write a good prompt. If you find yourself asking for follow-ups or providing clarifications a lot, consider putting more effort into upfront planning.
4. Don t Test The AI: If you know that a task will need the AI to overcome a certain challenge, or if you know about particularly challenging part involved by a task, DON'T hide that information from the AI.    
5. Explicitly Tell The AI About Tools It Should Use
```
